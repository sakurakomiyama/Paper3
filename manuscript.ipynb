# Introduction
## Predator-prey interactions
Understanding predator–prey interactions is essential for improving our understanding of marine ecosystems.
Fine-scale observations are important for resolving these interactions.
\
\
\

## Fine scale predator-prey interactions
Relatively few studies have documented concurrent, fine-scale predator–prey interactions in the field [@nottestad2002].
Simultaneous observations of marine predators and prey fish remain challenging because predators may avoid large research vessels, and predator–prey interactions may occur outside their sampling coverage. 
\
\
\

## Potential aid from USVs
To compensate for the constraints of vessel-based surveys, much attention has been paid to drone technology.
The opportunistic Saildrone observations of dolphins and seabirds provide an opportunity to explore fine scale predator-prey interactions.
